# Buddy Builder

Run the cell below (Shift+Enter) to start Buddy Builder; AppMode on nanoHUB runs it for you.

Start Jupyter from the project's environment so this kernel has the dependencies, e.g. on macOS:
`conda activate buddy`, `cd ~/Downloads/xtalligand`, `python -m notebook app.ipynb` (see README.md).
If the builder was already running, use **Kernel → Restart** before running the cell again.

```python
import sys
from pathlib import Path

# Prefer the copy installed by `make install` (lib/python), then the source tree.
here = Path.cwd()
for candidate in (here / "lib" / "python", here / "src"):
    if (candidate / "nanobuilder").is_dir():
        sys.path.insert(0, str(candidate))
        break

try:
    from nanobuilder.gui import BuilderApp
except ImportError as exc:  # usually Jupyter was started from a different Python
    raise ImportError(
        f"{exc}\n\nThis notebook's kernel runs {sys.executable} (Python "
        f"{sys.version.split()[0]}), which lacks the builder's dependencies. Stop this "
        "Jupyter server, activate the project's environment (conda activate buddy, "
        "or the .venv) and start Jupyter from it: python -m notebook app.ipynb "
        "(see README.md)."
    ) from exc

app = BuilderApp()
app.show()
```
